# Simple Linear Regression — BMI vs Life Expectancy

**What I wanted to find out:** can a country's average BMI predict its life expectancy?

Simple linear regression fits a straight line `y = b + w·x` with **one** input feature. The model picks `w` and `b` so that the squared errors between the line and the data points are as small as possible.

**Note:** scikit-learn needs `X` as a 2-D table (rows × features). That's why I used `[['BMI']]` (double brackets) and not `['BMI']`.

```python
import pandas as pd
from sklearn.linear_model import LinearRegression

# Load the data
bmi_life_data = pd.read_csv("/content/sample_data/bmi_and_life.csv")

# Make and fit the linear regression model
bmi_life_model = LinearRegression()
bmi_life_model.fit(bmi_life_data[['BMI']].values, bmi_life_data[['Life expectancy']].values)

# Make a prediction using the model
laos_life_exp = bmi_life_model.predict([[26.47563]])
print('The life expectancy of a person with a BMI of 26.47563', laos_life_exp)
```

## What I understood
- The model predicts a life expectancy of about **73.9 years** for a BMI of 26.48.
- This shows an **association** only. BMI is linked with many other factors (income, healthcare), so the model doesn't mean BMI *causes* a change in life expectancy.
- With only one feature, the model can't explain everything. Next, I tried multiple and polynomial regression.
