```python
# TODO: Add import statements
import pandas as pd
from sklearn.linear_model import LinearRegression

# Assign the Pandas dataframe to this variable.
# Load the data
bmi_life_data = pd.read_csv("/content/sample_data/bmi_and_life.csv")

# Make and fit the linear regression model
#TODO: Fit the model and Assign it to bmi_life_model
bmi_life_model = LinearRegression()
bmi_life_model.fit(bmi_life_data[['BMI']].values, bmi_life_data[['Life expectancy']].values)

# Make a prediction using the model
# TODO: Predict life expectancy for a BMI value of 23.47563 and print the result
laos_life_exp = bmi_life_model.predict([[26.47563]])
print('The life expectancy of a person with a BMI of 23.47563', laos_life_exp)
```
