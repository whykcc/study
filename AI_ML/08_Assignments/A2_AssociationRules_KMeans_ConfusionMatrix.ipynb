# Question 1 — Association Rules: Learning-Resource Use
**20 marks • Suggested time: 20 minutes**

A learning designer reviews the revision-week activity of **150 learners** in an online history course. Let **V** mean “accessed video recaps” and **M** mean “viewed concept maps”. The records show:

| Resource use | Number of learners |
|---|---:|
| Accessed video recaps (V) | 60 |
| Viewed concept maps (M) | 75 |
| Used both V and M | 42 |

The totals for V and M **include** the learners who used both. Some learners may have used neither resource. Each learner is counted once in the dataset. Analyse the association rule **$V\to M$** using these observational records.


### Q1(a) — 8 marks

Calculate the **support** and **confidence** of $V\to M$. Show the formula, the numerator, the denominator and the result for each measure. Express both final results as percentages.


**Your answer Q1(a)**

Support:
$$
\mathrm{Support}(V\to M)=\frac{n(V\cap M)}{N}
=\frac{42}{150}=0.28=28.00\%
$$

Confidence:
$$
\mathrm{Confidence}(V\to M)=\frac{n(V\cap M)}{n(V)}
=\frac{42}{60}=0.70=70.00\%
$$


### Q1(b) — 6 marks

Calculate the **lift** of $V\to M$. Show the overall proportion of learners who viewed concept maps, then use it in your calculation. Interpret the lift in the context of these learners. Use proportions, not whole-number percentages, in the division.


**Your answer Q1(b)**

Overall proportion who viewed concept maps:
$$
P(M)=\frac{n(M)}{N}=\frac{75}{150}=0.50
$$

Lift:
$$
\mathrm{Lift}(V\to M)=\frac{\mathrm{Confidence}(V\to M)}{P(M)}
=\frac{0.70}{0.50}=1.40
$$

Interpretation: The lift is 1.40. 70% of learners who watched video recaps also viewed concept maps, while only 50% of all learners viewed concept maps. So learners who watched video recaps were 1.40 times as likely to view concept maps. Because the lift is more than 1, there is a positive association between the two resources.


### Q1(c) — 6 marks

A teacher concludes: “Accessing video recaps makes learners view concept maps.” Evaluate this statement in **2–3 sentences**. Refer to what the association rule establishes, what it does not establish, and **one plausible alternative explanation** for the pattern.


**Your answer Q1(c)**
The lift of 1.40 can only prove that learners often use the two resources together. The data show this pattern, but we cannot prove that watching video recaps makes learners view concept maps. Maybe some learners viewed concept maps at the start of the lessons, and at the end they watched video recaps because of the exam.

### Q2(d) — 4 marks

State the centroid-update rule and explain why **both** clusters need new centroids when L5 moves. Then use your total WCSS values to state whether this step improves overall cluster compactness, and explain why this comparison alone does **not** prove that the clustering is globally optimal.


**Your answer Q2(d)**

Centroid-update rule: The new centroid of each cluster is the mean of all the points in that cluster. Its x-coordinate is the sum of x divided by the number of learners, and its y-coordinate is the sum of y divided by the number of learners.

Why both centroids change: When L5 moves, U loses one point and V gains one point. So the count and the sums of both clusters change, and both centroids must be calculated again. $C_U$ moves from (2.80, 2.80) to (2.00, 2.00), and $C_V$ moves from (8.00, 6.00) to (7.50, 6.00).

WCSS comparison and interpretation: The total WCSS drops from 41.60 to 19.00. This means the learners are now closer to their own centroids, so this step makes the clusters more compact. Most of the drop comes from U, because L5 was far from the other learners in U.

Why global optimality is not established: This comparison only shows that this step is better than the step before it. K-means can stop at a local minimum, and the result depends on the starting centroids. We did not check all the other possible ways to group the eight learners, so there may still be another grouping with a lower WCSS.

# Question 3 — Confusion Matrix and Classification Measures

In the table below, **rows are predictions** and **columns are actual outcomes**.

Evaluate the model for the positive class. An alert corresponds to a prediction of non-completion.

Positive class = Non-completion. Rows are predictions and columns are actual outcomes.

| Quantity | Number of learners |
|---|---|

### Q3(b) — 12 marks

Calculate **precision**, **recall** and the **F1 score** for the positive class. For each measure, show its formula, substitute the appropriate counts and show the result. Keep full precision until the final step. **4 marks per measure.**


**Your answer Q3(b)**

Precision:
$$
P=\frac{TP}{TP+FP}=\frac{42}{42+14}=\frac{42}{56}=0.7500
$$

Recall:
$$
R=\frac{TP}{TP+FN}=\frac{42}{42+18}=\frac{42}{60}=0.7000
$$

F1 score (use either equivalent form):
$$
F_1=\frac{2PR}{P+R}
=\frac{2TP}{2TP+FP+FN}=\frac{2(42)}{2(42)+14+18}=\frac{84}{116}=0.7241
$$

### Q3(c) — 6 marks

Explain the **false-negative** and **false-positive** errors in this module-completion setting. For each error, identify what the model predicted, what actually happened and how many learners were affected. Do not give only generic definitions.


**Your answer Q3(c)**

False negatives: The model predicted that these learners would complete the module, but they actually did not complete it. This happened to 18 learners. They did not get an alert, so the teacher may not give them extra help in time.

False positives: The model predicted that these learners would not complete the module, but they actually completed it. This happened to 14 learners. They got an alert that they did not need, so the teacher may spend extra time following them up.

### Q3(d) — 3 marks

Interpret your **recall** result for learners who actually did not complete the module. Include the numbers detected and missed, and state **one implication** for an educator deciding how to follow up learners.


**Your answer Q3(d)**

The recall is 0.7000 (70.00%). Among the 60 learners who actually did not complete the module, the model detected 42 of them, but it missed 18 learners. This means about 3 in 10 learners who needed help did not get an alert. For an educator, the alert list should not be the only way to find learners at risk. The teacher should also check other signs, such as low activity or late work, so that the missed learners can still get support.
