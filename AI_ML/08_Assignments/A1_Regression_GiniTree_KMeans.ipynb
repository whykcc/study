### Q2(a)

Consider the binary split **Attendance = Low** versus **Attendance = High**. Calculate the Yes/No proportions and Gini impurity in each child node. Then calculate the **weighted Gini impurity** of the split. Show all working.

**Your answer Q2(a)**

| Child node | Number of Yes | Number of No | $p_{\mathrm{Yes}}$ | $p_{\mathrm{No}}$ | Gini |
|---|---|---|---|---|---|

| Attendance = High | 0 | 4 | 0 | 1 | 0.0000 |

$$

$$

$$

# Question 3 — K-means and Centroid Calculations


A teacher explores groups of students using two practice scores, both measured on a **0–10 scale**. Use the raw coordinates as supplied. $K=3$, and the initial centroids are:

$$
C_1=(1,8),\qquad C_2=(5,7),\qquad C_3=(1,1).
$$

| Student | Mathematics score $x$ | Reading score $y$ |
|---|---:|---:|
| A | 1 | 8 |
| B | 2 | 3 |
| C | 8 | 2 |
| D | 5 | 7 |
| E | 7 | 4 |
| F | 7 | 3 |
| G | 1 | 1 |
| H | 4 | 9 |

### Q3(a)

After the initial centroids are set, describe the **assignment**, **centroid-update**, **repetition** and **stopping** steps of K-means.

**Your answer Q3(a)**

1. Assignment: Assign each student to the closest centroid.
2. Centroid update: Move each centroid to the center of the students assigned to it (the mean of their x and y coordinates).
3. Repetition: Loop to repeat the assignment and centroid-update steps using the new centroids.
4. Stopping: Stop when the centroids no longer change or the maximum number of iterations is reached.

### Q3(b)

Calculate the **Euclidean distance** from Student $B=(2,3)$ to each **initial** centroid. Show the working and identify the cluster to which B should be assigned.

**Your answer Q3(b)**

$$
d(B,C_1)=\sqrt{(x_B-x_{C_1})^2+(y_B-y_{C_1})^2}=\sqrt{(2-1)^2+(3-8)^2}=\sqrt{1+25}=\sqrt{26}\approx 5.10
$$
$$
d(B,C_2)=\sqrt{(x_B-x_{C_2})^2+(y_B-y_{C_2})^2}=\sqrt{(2-5)^2+(3-7)^2}=\sqrt{9+16}=\sqrt{25}=5.00
$$
$$
d(B,C_3)=\sqrt{(x_B-x_{C_3})^2+(y_B-y_{C_3})^2}=\sqrt{(2-1)^2+(3-1)^2}=\sqrt{1+4}=\sqrt{5}\approx 2.24
$$

Assign B to cluster C3 because its distance to C3 (2.24) is the smallest of the three distances (5.10, 5.00, 2.24), so C3 is the nearest centroid.

Calculate the **updated coordinates of all three centroids**, showing the mean calculation for each coordinate. Perform **one centroid-update step only**; do not reassign students afterwards.

**Your answer Q3(c)**

General mean-coordinate pattern: $C=\left(\frac{\sum_{i=1}^{n}x_i}{n},\frac{\sum_{i=1}^{n}y_i}{n}\right)$, where $n$ is the number of students in that cluster.

$$
C_1=\left(\frac{1}{1},\frac{8}{1}\right)=(1.00,\,8.00)

### Q3(d)

Explain **one concern** that may lead to a poor K-means result and suggest **one corresponding practical response**. No further numerical calculation is required.

**Your answer Q3(d)**

Concern and why it matters: K-means depends on where the initial centroids are placed. If the centroids are placed randomly or poorly, the algorithm may get stuck in a poor result, and the clusters do not go to the correct place. This matters because the teacher may group students wrongly and give them the wrong support.

Corresponding practical response: Run the K-means algorithm more than once with different initial centroids, and choose the best result, which is the one with the smallest total distance between the students and their centroids (the lowest WCSS).
